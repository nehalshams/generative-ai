```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
from pathlib import Path
from dotenv import load_dotenv
```

```python
from langchain_docling.loader import DoclingLoader
```

```python
cwd = Path.cwd()
cwd = cwd.parent if cwd.name == "rag" else cwd
db_path = cwd / "db"
db_path

file_path = cwd / "Documents" / "beyond-chatbots-ai-agents-next-real-shift.pdf"
print(file_path)
```

```python
load_dotenv()
```

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader(file_path=file_path)
docs = loader.load()
docs
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embeddings
```

```python
vectorstore = Chroma.from_documents(
    collection_name = 'beyond_the_chatbots',
    documents=docs,
    persist_directory=str(db_path), 
    embedding=embeddings
)
vectorstore
```

```python
query = "How do AI agents use tools and memory?"
query
```

```python
result = vectorstore.similarity_search(query, k=2)
result
```

```python
search_with_score = vectorstore.similarity_search_with_score(query, k=2)
search_with_score
```
